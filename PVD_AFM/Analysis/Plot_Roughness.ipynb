# Visualize Roughness Data

## Preparation

### Imports

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
```

### Parameters

```python
DATA_PATH    = "../Data/PVD_SR_by_WaferPos.csv"
MEASURES     = 'Sq_pm', 'Sa_pm', 'Ssk', 'Sp_nm', 'Sv_nm', 'Sz_nm'
NUM_MEAS     = len(MEASURES)

FIG_ROOT_DIR = "../Figures/"
```

### File Stuff

```python
# Load the CSV data
SR_by_WaferPos = pd.read_csv(DATA_PATH)

# Create Figures Dir
os.makedirs(FIG_ROOT_DIR, exist_ok=True)
```

## Visualize by spot on wafer

```python
def plot_roughness_by_position(data):
    """
    Plots surface roughness by wafer position.

    Parameters:
    data (DataFrame): DataFrame containing 'Wafer_Position' and 'Surface_Roughness' columns.
    """

    unique_wafers       = data['Wafer'].unique()
    unique_electrodes   = data['Which_Electrode'].unique()
    unique_positions    = data['Wafer_Pos'].unique()

    pltdims = len(unique_wafers) * len(unique_electrodes), len(MEASURES)
    fig, ax = plt.subplots(pltdims[0], pltdims[1], figsize=(pltdims[1]*4, pltdims[0]*2.5), squeeze=True)
    ax      = ax.reshape(pltdims)

    fig.suptitle('Surface Roughness by Wafer Position', fontsize=16)
    ith_row = 0
    for ith_wafer, wafer in enumerate(unique_wafers):
        for ith_electrode, electrode in enumerate(unique_electrodes):
            plt_row_data = data[(data.Wafer == wafer) & (data.Which_Electrode == electrode)]
            for ith_col, measure in enumerate(MEASURES):
                ax[ith_row, ith_col].set_xlabel("Wafer Position")
                ax[ith_row, ith_col].set_ylabel(measure)
                ax[ith_row, ith_col].set_title(f'{measure}, Wafer {wafer}, {electrode} Electrode')
                ax[ith_row, ith_col].scatter(plt_row_data['Wafer_Pos'], plt_row_data[measure], color='blue', alpha=0.7)
                ax[ith_row, ith_col].grid(True)
            ith_row += 1

    plt.tight_layout()
    plt.savefig(f"{FIG_ROOT_DIR}/Surface_Roughness_by_Wafer_Position.png", dpi=300)
    plt.close()
```

```python
plot_roughness_by_position(SR_by_WaferPos)
```
