```python
import numpy as np
```

```python
EPS_0 = 8.854e-12
EPS_R = {
    "HfO2" : lambda f: 18 if f == 10e3 else 0 and print("I only know Eps_R at 10kHz!")
}

DIELECTRIC       = "HfO2"
MEASUREMENT_FREQ = 10e4

R = np.linspace(100, 1000, 10) * 1e-6 * .5  # Radii of circular parallel plate capacitors
A = np.pi * R * R                           # Surface area of overlapping plates

S = 20e-9
```

```python
def format_suffix(f):
    if f == 0:
        return f"{0.0:8.2f}  "
    if f <= 1:
        suffix = " munpf"
        i = 0
        while f < 1:
            f *= 1e3
            i += 1
        
        return f"{f:8.2f} {suffix[i]}"

    elif f >= 1:
        suffix = " KMGTPE"
        i = 0
        while f >= 1e3:
            f *= 1e-3
            i += 1
        
        return f"{f:8.2f} {suffix[i]}"
```

```python
eps_test    = EPS_R[DIELECTRIC](MEASUREMENT_FREQ)
C_expected  = EPS_0 * eps_test * A / S

print(f"Max separation to radius ratio: {np.max(S / R) * 100:.3f}%", end="")
if np.max(S / R) > 0.01:
    print(f" - Warning: Parallel Plate Assumption not Satisfied!")
else:
    print(" - Parallel Plate Assumption Satisfied")
print(f"eps_r of {DIELECTRIC} at {format_suffix(MEASUREMENT_FREQ)}Hz : {eps_test:.2f}")
print()

print("   Diameter    C_expected  ")
print(" ---------------------------")
for r, c in zip(R, C_expected):
    print(f"{format_suffix(2*r)}m   {format_suffix(c)}F")
```
